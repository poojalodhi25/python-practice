```python
n = 12345
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

print(reverse)
```

```python
n = 121
original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
n = 153
original = n
count = 0

# Count digits
temp = n

while temp > 0:
    count += 1
    temp = temp // 10

# Calculate Armstrong sum
temp = n
total = 0

while temp > 0:
    digit = temp % 10
    total = total + digit ** count
    temp = temp // 10

# Check
if original == total:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```
