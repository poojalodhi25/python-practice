```python
def my_decorator(func):
  def wrapper():
    print('*****************')
    func()
    print('******************')
  return wrapper

def hello():
   print('hello')

def display():
  print('hello nitish')
a =my_decorator(hello)
a()

b = my_decorator(display)
b()
```

```python
import time

def timer(func):

    def wrapper():
        start = time.time()

        func()

        print('time taken by', func.__name__, time.time() - start, 'secs')

    return wrapper


@timer
def hello():
    print('hello world')
    time.sleep(2)


@timer
def display():
    print('display something')
    time.sleep(1)


hello()
```
