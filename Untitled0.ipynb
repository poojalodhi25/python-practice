```python
def is_even(num):
  if type(num)==int:
    if num % 2 == 0:
        return True
    else:
        return False
  else:
    return 'please enter integer number'
```

```python
#function
#function_name(input)
for i in range(1,11):
  x = is_even(i)
  print(x)
```

```python
#args
# allow us to pass a variable number of non key argu to a fun
def multiply(*args):
  product =1

  for i in args:
    product = product*i
    print(args)
  return product

```

```python
#kwargs

#allow us to pass any num of kw argu
#kw argu mean that they contain a key-value pair, like a python dictiornary
def display(**kwargs):
  for(key, value) in kwargs.items():
    print ( key,'->', value)
```
