```python
def is_even(num):
  if type(num)==int:
    if num % 2 == 0:
        return True
    else:
        return False
  else:
    return 'please enter integer number'
```

```python
#function
#function_name(input)
for i in range(1,11):
  x = is_even(i)
  print(x)
```
